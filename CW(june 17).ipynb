```python
import numpy as np
import pandas as pd
from dateutil.parser import parse
```

```python
df= pd.DataFrame({
    "name":["Abel","Sara",None,"John"],
    "age":[20,np.nan,25,30],
    "salary":[4000,4500,5000,45000],
    "join_date":[
        "2026-01-01",
        "01/02/2026",
        "March 10,2026",
        "2026.04.15"
    ]})
```

```python
print(df)
```

```python
df["name"] = df["name"].fillna("Unknown")
print(df)
```

```python
for column in df.columns:
    if df[column].dtype == "float64" or df[column].dtype == "int64":
        df[column].fillna(df[column].mean(), inplace=True)

print (df)
```

```python

def parse_date(date_str):
    try:
        return parse(date_str, dayfirst=False)
    except ValueError:
        return pd.NaT

df['join_date'] = df['join_date'].apply(parse_date)

df['join_date'] = df['join_date'].dt.strftime('%Y-%m-%d')

print("Result")
print(df)
```

```python
Q1 = df["salary"].quantile(0.25)
Q3 = df["salary"].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df["Salary_Outlier"] = df["salary"].apply(
    lambda x: "Outlier"
    if x < lower_bound or x > upper_bound
    else "Normal"
)


print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)
print("Lower Bound =", lower_bound)
print("Upper Bound =", upper_bound)

print("\nFinal DataFrame")
print(df)
```
