```python
import pandas as pd
import numpy as np

# =========================
# PART 1: LOAD DATASET
# =========================

url = "employee_data.csv"  
df = pd.read_csv(url)

print("FIRST 10 ROWS")
print(df.head(10))

print("\nLAST 10 ROWS")
print(df.tail(10))

print("\nROWS AND COLUMNS")
print(df.shape)

print("\nDATA TYPES")
print(df.dtypes)

print("\nSUMMARY STATISTICS")
print(df.describe(include="all"))

print("\nCATEGORICAL COLUMNS")
categorical_cols = df.select_dtypes(include="object").columns
print(categorical_cols)

print("\nNUMERICAL COLUMNS")
numerical_cols = df.select_dtypes(include=np.number).columns
print(numerical_cols)
```

```python
# =========================
# PART 2: MISSING VALUES
# =========================

print("\nMISSING VALUES")
print(df.isnull().sum())

print("\nCOLUMNS WITH MISSING VALUES")
print(df.columns[df.isnull().any()])

df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
df["City"] = df["City"].fillna(df["City"].mode()[0])
df["Performance_Score"] = df["Performance_Score"].fillna(df["Performance_Score"].mean())

print("\nMISSING VALUES AFTER FILLING")
print(df.isnull().sum())
```

```python
# =========================
# PART 3: DUPLICATE RECORDS
# =========================

print("\nDUPLICATE ROWS")
print(df[df.duplicated()])

print("\nNUMBER OF DUPLICATES")
print(df.duplicated().sum())

df = df.drop_duplicates()

print("\nDUPLICATES AFTER REMOVING")
print(df.duplicated().sum())
```

```python
# =========================
# PART 4: INCONSISTENT DATA
# =========================

df["Gender"] = df["Gender"].str.strip().str.title()
df["Status"] = df["Status"].str.strip().str.title()
df["Email"] = df["Email"].str.strip().str.lower()

df["Department"] = df["Department"].str.strip().str.title()

department_map = {
    "It": "Information Technology",
    "Info Tech": "Information Technology",
    "Information Technology": "Information Technology",
    "Hr": "Human Resources",
    "Human Resource": "Human Resources",
    "Human Resources": "Human Resources"
}

df["Department"] = df["Department"].replace(department_map)

print("\nSTANDARDIZED DATA SAMPLE")
print(df.head())
```

```python
# =========================
# PART 5: DATA ENTRY ERRORS
# =========================

df.loc[df["Age"] < 0, "Age"] = np.nan
df.loc[df["Age"] > 100, "Age"] = np.nan
df["Age"] = df["Age"].fillna(df["Age"].mean())

df.loc[df["Salary"] < 0, "Salary"] = np.nan
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

df.loc[df["Performance_Score"] < 0, "Performance_Score"] = np.nan
df.loc[df["Performance_Score"] > 100, "Performance_Score"] = np.nan
df["Performance_Score"] = df["Performance_Score"].fillna(df["Performance_Score"].mean())
```

```python
# =========================
# PART 6: OUTLIER DETECTION
# =========================

highest_salary = df["Salary"].max()
lowest_salary = df["Salary"].min()

print("\nHIGHEST SALARY")
print(highest_salary)

print("\nLOWEST SALARY")
print(lowest_salary)

Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

salary_outliers = df[(df["Salary"] < lower_bound) | (df["Salary"] > upper_bound)]

print("\nSALARY OUTLIERS")
print(salary_outliers)

print("\nEMPLOYEES EARNING ABOVE 100,000")
print(df[df["Salary"] > 100000])

avg_salary_with_outliers = df["Salary"].mean()
avg_salary_without_outliers = df[
    (df["Salary"] >= lower_bound) & (df["Salary"] <= upper_bound)
]["Salary"].mean()

print("\nAVERAGE SALARY WITH OUTLIERS")
print(avg_salary_with_outliers)

print("\nAVERAGE SALARY WITHOUT OUTLIERS")
print(avg_salary_without_outliers)
```

```python
# =========================
# PART 7: FEATURE ENGINEERING
# =========================

def age_group(age):
    if age <= 30:
        return "Young"
    elif age <= 45:
        return "Mid"
    else:
        return "Senior"

df["Age_Group"] = df["Age"].apply(age_group)

df["Salary_Category"] = pd.cut(
    df["Salary"],
    bins=[0, 50000, 100000, float("inf")],
    labels=["Low", "Medium", "High"]
)

def experience_level(exp):
    if exp <= 2:
        return "Junior"
    elif exp <= 5:
        return "Mid"
    else:
        return "Senior"

df["Experience_Level"] = df["Experience"].apply(experience_level)

df["Bonus_Eligible"] = np.where(
    (df["Performance_Score"] > 85) & (df["Experience"] > 3),
    "Yes",
    "No"
)

print("\nFEATURE ENGINEERING SAMPLE")
print(df.head())
```

```python
# =========================
# PART 8: EXPLORATORY DATA ANALYSIS
# =========================

print("\nTOTAL EMPLOYEES")
print(len(df))

print("\nMALES AND FEMALES")
print(df["Gender"].value_counts())

print("\nDEPARTMENT WITH MOST EMPLOYEES")
print(df["Department"].value_counts().idxmax())

print("\nBRANCH WITH MOST EMPLOYEES")
print(df["Branch"].value_counts().idxmax())

print("\nCITY WITH MOST EMPLOYEES")
print(df["City"].value_counts().idxmax())

print("\nAVERAGE SALARY BY DEPARTMENT")
print(df.groupby("Department")["Salary"].mean())

print("\nAVERAGE SALARY BY BRANCH")
print(df.groupby("Branch")["Salary"].mean())

print("\nAVERAGE PERFORMANCE SCORE BY DEPARTMENT")
print(df.groupby("Department")["Performance_Score"].mean())

print("\nTOP 10 HIGHEST PAID EMPLOYEES")
print(df.sort_values(by="Salary", ascending=False).head(10))

print("\nTOP 10 MOST EXPERIENCED EMPLOYEES")
print(df.sort_values(by="Experience", ascending=False).head(10))

print("\nRELATIONSHIP BETWEEN EXPERIENCE AND SALARY")
print(df[["Experience", "Salary"]].corr())
```

```python
# =========================
# PART 9: BUSINESS QUESTIONS
# =========================

print("\nDEPARTMENT THAT COSTS THE COMPANY THE MOST")
print(df.groupby("Department")["Salary"].sum().idxmax())

print("\nBRANCH THAT PERFORMS BEST")
print(df.groupby("Branch")["Performance_Score"].mean().idxmax())

print("\nEMPLOYEES WHO DESERVE BONUSES")
print(df[df["Bonus_Eligible"] == "Yes"])

print("\nIS THERE A RELATIONSHIP BETWEEN EXPERIENCE AND SALARY?")
correlation = df["Experience"].corr(df["Salary"])
print(correlation)

if correlation > 0.5:
    print("There is a strong positive relationship.")
elif correlation > 0.2:
    print("There is a weak positive relationship.")
elif correlation < -0.2:
    print("There is a negative relationship.")
else:
    print("There is little or no relationship.")

print("\nGENDER IMBALANCE")
print(df["Gender"].value_counts())

print("\nCITY WITH LARGEST WORKFORCE")
print(df["City"].value_counts().idxmax())

print("\nSUSPICIOUS SALARY VALUES")
print(df[(df["Salary"] < 10000) | (df["Salary"] > 100000)])

print("\nDEPARTMENT THAT MAY REQUIRE ADDITIONAL HIRING")
print(df["Department"].value_counts().idxmin())
```

```python
# =========================
# SAVE CLEANED DATA
# =========================

df.to_csv("cleaned_employee_data.csv", index=False)

print("\nCleaned file saved as cleaned_employee_data.csv")
```
