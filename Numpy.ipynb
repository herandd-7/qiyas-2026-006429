```python
arr = np.array([[-1, 2, 0, 4], [4, -0.5, 6, 0], [2.6, 0, 7, 8], [3, -7, 4, 2.0]])
arr2 = arr[:2, ::2]
print ("first 2 rows and alternate columns(0 and 2):\n", arr2)#review this later
arr3 = arr[[1, 1, 0, 3], [3, 2, 1, 0]] #review this later
print ("\nElements at indices (1, 3), (1, 2), (0, 1), (3, 0):\n", arr3)
```

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[4, 3], [2, 1]])
print ("Adding 1 to every element:", a + 1)
print ("\nSubtracting 2 from each element:", b - 2)
print ("\nSum of all array elements: ", a.sum())
print ("\nArray sum:\n", a + b)
```
