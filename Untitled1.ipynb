```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

base_url = "https://www.fanabc.com/english"

categories = {
    "Local News": "/category/local-news/",
    "Business": "/category/business/",
    "World News": "/category/worldnews/",
    "Sport": "/category/sport/",
    "Africa": "/category/africa/",
    "Opinion": "/category/opinion/",
}

headers = {
    "User-Agent": "Mozilla/5.0"
}

all_articles = []

for category, path in categories.items():
    url = base_url + path
    page = requests.get(url, headers=headers)
    soup = BeautifulSoup(page.text, "html.parser")

    articles = soup.find_all("article")

    for article in articles:
        title_tag = article.find("h2", class_="title")
        link_tag = title_tag.find("a") if title_tag else None

        title = link_tag.get_text(strip=True) if link_tag else ""
        article_url = link_tag["href"] if link_tag and link_tag.has_attr("href") else ""

        date_tag = article.find("span", class_="time")
        date = date_tag.get_text(strip=True) if date_tag else ""

        writer_tag = article.find("div", class_="post-meta")
        writer = writer_tag.get_text(" ", strip=True) if writer_tag else ""

        img = ""
        img_tag = article.find("a", class_="img-holder")
        if img_tag and img_tag.has_attr("style"):
            style = img_tag["style"]
            if "url(" in style:
                img = style.split('url("')[1].split('")')[0]

        content = ""

        if article_url:
            article_page = requests.get(article_url, headers=headers)
            article_soup = BeautifulSoup(article_page.text, "html.parser")

            content_div = article_soup.find("div", class_="entry-content clearfix single-post-content")

            if content_div:
                paragraphs = content_div.find_all("p")
                content = " ".join(p.get_text(" ", strip=True) for p in paragraphs)

        all_articles.append({
            "category": category,
            "source_url": url,
            "article_url": article_url,
            "title": title,
            "date": date,
            "writer": writer,
            "content": content,
        })

df = pd.DataFrame(all_articles)
df.to_csv("fanabc.csv", index=False)

print(df.head())
```
