```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_excel("COVID-19-geographic-disbtribution-worldwide.xlsx")
```

```python
# Keep African countries only
africa = df[df["continentExp"] == "Africa"]

# Plot average cases by month
sns.relplot(
    data=africa,
    kind="line",
    x="month",
    y="cases",
   
)

plt.title("Average Daily COVID Cases in Africa by Month")
plt.xlabel("Month")
plt.ylabel("Average Daily Cases")
plt.show()
```

```python
from dash import Dash, dcc, html, Input, Output
import pandas as pd
import plotly.express as px

#try by african countries next
df = pd.read_excel("COVID-19-geographic-disbtribution-worldwide.xlsx")


days_of_month = sorted(df["day"].dropna().unique())

app = Dash(__name__)

app.layout = html.Div([
    html.H1("COVID-19 Deaths by Day of Month"),

    dcc.Dropdown(
        id="dropdown",
        options=[{"label": str(day), "value": day} for day in days_of_month],
        value=days_of_month[0],
        clearable=False
    ),

    dcc.Graph(id="graph")
])

@app.callback(
    Output("graph", "figure"),
    Input("dropdown", "value")
)


def update_bar_chart(selected_day):

    filtered_df = df[df["day"] == selected_day]

    fig = px.bar(
        filtered_df,
        x="countriesAndTerritories",
        y="deaths",
        title=f"Deaths on Day {selected_day}",
        labels={
            "countriesAndTerritories": "Country",
            "deaths": "Deaths"
        }
    )

    return fig

app.run(debug=True)
```
