```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import re

base_url = "https://www.fanamc.com/english"

categories = {
    "Local News": "/category/local-news/",
    "Business": "/category/business/",
    "World News": "/category/worldnews/",
    "Sport": "/category/sport/",
    "Africa": "/category/africa/",
    "Opinion": "/category/opinion/",
}

headers = {
    "User-Agent": "Mozilla/5.0"
}

all_articles = []
seen_urls = set()

for category, path in categories.items():
    source_url = base_url + path
    page = requests.get(source_url, headers=headers)
    soup = BeautifulSoup(page.text, "html.parser")

    articles = soup.find_all("article")

    for article in articles:
        title_tag = article.find("h2", class_="title")
        link_tag = title_tag.find("a") if title_tag else None

        if not link_tag:
            continue

        title = link_tag.get_text(strip=True)
        article_url = link_tag.get("href", "")

        # skip duplicates
        if article_url in seen_urls:
            continue
        seen_urls.add(article_url)

        # image from category page
        img = ""
        img_tag = article.find("a", class_="img-holder")

        if img_tag:
            style = img_tag.get("style", "")
            match = re.search(r'url\(["\']?(.*?)["\']?\)', style)

            if match:
                img = match.group(1)

        # go inside article page
        article_page = requests.get(article_url, headers=headers)
        article_soup = BeautifulSoup(article_page.text, "html.parser")

        date = ""
        writer = ""
        content = ""

        date_tag = article_soup.find("span", class_="time")
        if date_tag:
            date = date_tag.get_text(strip=True)

        writer_tag = article_soup.find("span", class_="post-author")
        if writer_tag:
            writer = writer_tag.get_text(strip=True)

        content_div = article_soup.find("div", class_="entry-content")
        if content_div:
            paragraphs = content_div.find_all("p")
            content = " ".join(p.get_text(strip=True) for p in paragraphs)

        all_articles.append({
            "category": category,
            "source_url": source_url,
            "article_url": article_url,
            "title": title,
            "date": date,
            "writer": writer,
            "content": content,
            "img": img
        })

df = pd.DataFrame(all_articles)
df.to_csv("fanabc_art.csv", index=False, encoding="utf-8-sig")

print(df)
print("Total articles:", len(df))
```
