```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots()
fruits = ['apple', 'blueberry', 'Avocado', 'orange']
counts = [40, 75, 30, 55]
bar_labels = ['red', 'blue', 'green', 'orange']
bar_colors = ['tab:red', 'tab:blue', 'tab:green', 'tab:orange']
ax.bar(fruits, counts, label=bar_labels, color=bar_colors)
ax.set_ylabel('Supply')
ax.set_title('Fruit by kind and color')
ax.legend(title='Color')
plt.show() #you can specify image size
```

```python
import numpy as np
dept = ('CS', 'Biology', 'IS', 'CE')
sex_counts = {'Male': np.array([73, 34, 61, 56]), 'Female': np.array([57, 34, 58,29])}
width = 0.75 # the width of the bars: can also be len(x) sequence
fig, ax = plt.subplots()
bottom = np.zeros(4)
for sex, sex_count in sex_counts.items():
    p = ax.bar(dept, sex_count, width, label=sex, bottom=bottom)
    bottom += sex_count
    ax.bar_label(p, label_type='center')
ax.set_title('Number of student by sex for different courses')
ax.legend()
plt.show()
```

```python
df = pd.DataFrame(data)

pivoted_data = df.pivot_table(
    index='Year',
    values='CGPA',
    aggfunc='mean'
)

print(pivoted_data)
```

```python
import pandas as pd
import matplotlib.pyplot as plt


data = {
    "Year": [
        "Year I","Year III","Year I","Year IV","Year I",
        "Year I","Year III","Year I","Year I","Year III",
        "Year I","Year III","Year III"
    ],
    "CGPA": [
        2.9,3.4,3.9,3.0,2.9,
        3.4,3.9,1.5,2.9,3.4,
        3.9,3.0,2.8
    ],
    "SGPA": [
        2.7,3.2,3.7,2.8,2.7,
        3.6,3.7,1.5,2.7,3.2,
        3.7,2.8,2.6
    ]
}

df = pd.DataFrame(data)


pivoted_data = df.pivot_table(
    index="Year",
    values=["CGPA", "SGPA"],
    aggfunc="mean"
)

print(pivoted_data)


pivoted_data.plot(kind="bar")

plt.title("Average CGPA and SGPA by Year")
plt.xlabel("Year")
plt.ylabel("Average GPA")
plt.xticks(rotation=0)
plt.legend(["CGPA", "SGPA"])

plt.show()
```
