```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load Excel file
df = pd.read_excel("COVID-19-geographic-disbtribution-worldwide.xlsx")

# Check column names
print(df.columns)

# Filter Ethiopia only
ethiopia = df[df["countriesAndTerritories"] == "Ethiopia"].copy()

# Convert date column to real dates
ethiopia["dateRep"] = pd.to_datetime(
    ethiopia["dateRep"],
    dayfirst=True
)

# Create month column
ethiopia["month"] = ethiopia["dateRep"].dt.month

# Optional: create month name
ethiopia["month_name"] = ethiopia["dateRep"].dt.month_name()

# Plot average daily COVID cases by month
sns.relplot(
    data=ethiopia,
    kind="line",
    x="month",
    y="cases",
    errorbar=("ci", 95),
    marker="o"
)

plt.title("Average Daily COVID Cases by Month in Ethiopia")
plt.xlabel("Month")
plt.ylabel("Average Daily Cases")
plt.show()
```

```python
import pandas as pd

# Convert the "dateRep" column from text (string) into actual datetime objects.
# This allows us to perform date calculations, extract months, years, days, etc.
# dayfirst=True is used because the dates are stored as DD/MM/YYYY.
ethiopia["dateRep"] = pd.to_datetime(
    ethiopia["dateRep"],
    dayfirst=True
)

# Create a new column containing only the month number (1-12).
# This is useful when grouping data by month to find monthly totals or averages.
# Example:
# 24/03/2020 -> 3
# 15/12/2020 -> 12
ethiopia["month"] = ethiopia["dateRep"].dt.month

# Create another column containing the month name instead of a number.
# This makes graphs and tables easier to read.
# Example:
# 3 -> March
# 12 -> December
ethiopia["month_name"] = ethiopia["dateRep"].dt.month_name()

# View the first few rows to confirm the new columns were created correctly.
print(ethiopia.head())
```
