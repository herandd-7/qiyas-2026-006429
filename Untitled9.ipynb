```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load data
df = pd.read_excel("COVID-19-geographic-disbtribution-worldwide.xlsx")

# Convert dates
df["dateRep"] = pd.to_datetime(
    df["dateRep"],
    dayfirst=True
)

# Keep African countries only
africa = df[df["continentExp"] == "Africa"].copy()

# Create month column
africa["month"] = africa["dateRep"].dt.month

# Plot average daily cases by month for African countries
sns.relplot(
    data=africa,
    kind="line",
    x="month",
    y="cases",
    errorbar=("ci", 95),
    marker="o"
)

plt.title("Average Daily COVID Cases by Month in Africa")
plt.xlabel("Month")
plt.ylabel("Average Daily Cases")
plt.show()
```
