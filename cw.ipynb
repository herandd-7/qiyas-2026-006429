```python
import pandas as pd
import matplotlib.pyplot as plt

# Create the dataset
data = {
    "Year": [
        "Year I","Year III","Year I","Year IV","Year I",
        "Year I","Year III","Year I","Year I","Year III",
        "Year I","Year III","Year III"
    ],
    "CGPA": [
        2.9,3.4,3.9,3.0,2.9,
        3.4,3.9,1.5,2.9,3.4,
        3.9,3.0,2.8
    ],
    "SGPA": [
        2.7,3.2,3.7,2.8,2.7,
        3.6,3.7,1.5,2.7,3.2,
        3.7,2.8,2.6
    ]
}

df = pd.DataFrame(data)

# Pivot table
pivoted_data = df.pivot_table(
    index="Year",
    values=["CGPA", "SGPA"],
    aggfunc="mean"
)

print(pivoted_data)

# Bar chart
pivoted_data.plot(kind="bar")

plt.title("Average CGPA and SGPA by Year")
plt.xlabel("Year")
plt.ylabel("Average GPA")
plt.xticks(rotation=0)
plt.legend(["CGPA", "SGPA"])

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Create the dataset
data = {
    "ID": [
        "ID-001", "ID-002", "ID-003", "ID-004", "ID-005",
        "ID-006", "ID-007", "ID-008", "ID-009", "ID-010",
        "ID-011", "ID-012", "ID-013"
    ],

    "Year": [
        "Year I", "Year III", "Year I", "Year IV", "Year I",
        "Year I", "Year III", "Year I", "Year I", "Year III",
        "Year I", "Year III", "Year III"
    ],

    "CGPA": [
        2.9, 3.4, 3.9, 3.0, 2.9,
        3.4, 3.9, 1.5, 2.9, 3.4,
        3.9, 3.0, 2.8
    ],

    "SGPA": [
        2.7, 3.2, 3.7, 2.8, 2.7,
        3.6, 3.7, 1.5, 2.7, 3.2,
        3.7, 2.8, 2.6
    ]
}

# Create DataFrame
df = pd.DataFrame(data)

# Create Pivot Table
pivoted_data = df.pivot_table(
    index="Year",
    values=["CGPA", "SGPA"],
    aggfunc="mean"
)

print(pivoted_data)

# Extract values
years = pivoted_data.index
cgpa = pivoted_data["CGPA"]
sgpa = pivoted_data["SGPA"]

# Create stacked bar chart
fig, ax = plt.subplots()

ax.bar(years, sgpa, label="SGPA")
ax.bar(years, cgpa, bottom=sgpa, label="CGPA")

ax.set_title("Average CGPA and SGPA by Year")
ax.set_xlabel("Year")
ax.set_ylabel("Average GPA")
ax.legend()

plt.show()
```
