```python
import pandas as pd
import matplotlib.pyplot as plt

# Create the dataset
data = {
    "ID": [
        "ID-001", "ID-002", "ID-003", "ID-004", "ID-005",
        "ID-006", "ID-007", "ID-008", "ID-009", "ID-010",
        "ID-011", "ID-012", "ID-013"
    ],

    "Year": [
        "Year I", "Year III", "Year I", "Year IV", "Year I",
        "Year I", "Year III", "Year I", "Year I", "Year III",
        "Year I", "Year III", "Year III"
    ],

    "CGPA": [
        2.9, 3.4, 3.9, 3.0, 2.9,
        3.4, 3.9, 1.5, 2.9, 3.4,
        3.9, 3.0, 2.8
    ],

    "SGPA": [
        2.7, 3.2, 3.7, 2.8, 2.7,
        3.6, 3.7, 1.5, 2.7, 3.2,
        3.7, 2.8, 2.6
    ]
}

# Create DataFrame
df = pd.DataFrame(data)

# Create Pivot Table
pivoted_data = df.pivot_table(
    index="Year",
    values=["CGPA", "SGPA"],
    aggfunc="mean"
)

print(pivoted_data)

# Extract values
years = pivoted_data.index
cgpa = pivoted_data["CGPA"]
sgpa = pivoted_data["SGPA"]

# Create stacked bar chart
fig, ax = plt.subplots()

ax.bar(years, sgpa, label="SGPA")
ax.bar(years, cgpa, bottom=sgpa, label="CGPA")

ax.set_title("Average CGPA and SGPA by Year")
ax.set_xlabel("Year")
ax.set_ylabel("Average GPA")
ax.legend()

plt.show()
```

```python
import seaborn as sns
import pandas as pd

# Load the dataset
tips = sns.load_dataset("tips")


# ==========================================
# COUNT
# How many records (tips) exist for each day?
# ==========================================
tip_count = tips.groupby("day")["tip"].count()
print("Count:")
print(tip_count)


# ==========================================
# MEAN (AVERAGE)
# Average tip amount for each day
# ==========================================
tip_mean = tips.groupby("day")["tip"].mean()
print("\nMean:")
print(tip_mean)


# ==========================================
# MEDIAN
# The middle tip value for each day
# Less affected by extreme values than the mean
# ==========================================
tip_median = tips.groupby("day")["tip"].median()
print("\nMedian:")
print(tip_median)


# ==========================================
# SUM
# Total amount of tips collected on each day
# ==========================================
tip_sum = tips.groupby("day")["tip"].sum()
print("\nSum:")
print(tip_sum)


# ==========================================
# MINIMUM
# Smallest tip given on each day
# ==========================================
tip_min = tips.groupby("day")["tip"].min()
print("\nMinimum:")
print(tip_min)


# ==========================================
# MAXIMUM
# Largest tip given on each day
# ==========================================
tip_max = tips.groupby("day")["tip"].max()
print("\nMaximum:")
print(tip_max)


# ==========================================
# STANDARD DEVIATION
# Measures how spread out the tips are
# Higher value = more variation in tip amounts
# ==========================================
tip_std = tips.groupby("day")["tip"].std()
print("\nStandard Deviation:")
print(tip_std)


# ==========================================
# ALL STATISTICS TOGETHER
# Quick summary table by day
# ==========================================
summary = tips.groupby("day")["tip"].agg([
    "count",
    "mean",
    "median",
    "sum",
    "min",
    "max",
    "std"
])

print("\nSummary:")
print(summary)
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the Excel file
df = pd.read_excel("COVID-19-geographic-disbtribution-worldwide.xlsx")

# Keep only African countries/territories
africa = df[df["continentExp"] == "Africa"].copy()

# Create a month column from dateRep
# Example: "24/06/2020" becomes month 6
africa["month"] = africa["dateRep"].str.split("/").str[1].astype(int)

# Group by month and add up all cases and deaths in Africa
monthly_data = africa.groupby("month")[["cases", "deaths"]].sum().reset_index()

# Plot cases line
sns.lineplot(
    data=monthly_data,
    x="month",
    y="cases",
    marker="o",
    label="Cases"
)

# Plot deaths line on the same graph
sns.lineplot(
    data=monthly_data,
    x="month",
    y="deaths",
    marker="o",
    label="Deaths"
)

# Add title and labels
plt.title("Monthly COVID Cases and Deaths in Africa")
plt.xlabel("Month")
plt.ylabel("Count")

plt.show()
```
