```python
import requests
from bs4 import BeautifulSoup

html_doc = requests.get('https://www.worldometers.info/coronavirus/')
soup = BeautifulSoup(html_doc.content, 'html.parser')

table_content = soup.find('table')
table_head = table_content.find('thead')
table_body = table_content.find('tbody')
table_row = table_body.find_all('tr')


    
```

```python
cells = table_row[8].find_all('td')

for i in range(len(cells)):
    print(i, ":", cells[i].text.strip())
```

```python
file = open("worldmeters.csv", "w", encoding="utf-8")

# Write headers
header_list = [
    "Country",
    "Total Cases",
     "New Cases",
    "Total Deaths",
     "New deaths",
    "Total Recovered",
     "New recovered",
     "Active cases",
     "serious/critical",
    "Tot Cases/1M pop",
     "Deaths/1M pop",
    "Total Tests",
     "Tests/1M pop",
    "Population",
     "Case evry X ppl",
    "Death evry X ppl",
    "Test evry X ppl",
     "New cases/1m pop",
     "New Deaths/1m pop",
     "Active cases/1m pop",
     
]

file.write(",".join(header_list))
file.write("\n")

# Loop through countries
for row in table_row[8:]:

    cells = row.find_all("td")
    
    row_data = []

    row_data.append(cells[1].text.strip())   # Country
    row_data.append(cells[2].text.strip())   # Total Cases
    row_data.append(cells[3].text.strip())   # New Cases
    row_data.append(cells[4].text.strip())   # Total Deaths
    row_data.append(cells[5].text.strip())   # New Deaths
    row_data.append(cells[6].text.strip())   # Total Recovered
    row_data.append(cells[7].text.strip())   # New Recovered
    row_data.append(cells[8].text.strip())   # Active Cases
    row_data.append(cells[9].text.strip())   # Serious/Critical
    row_data.append(cells[10].text.strip())  # Total Cases/1M pop
    row_data.append(cells[11].text.strip())  # Deaths/1M pop
    row_data.append(cells[12].text.strip())  # Total Tests
    row_data.append(cells[13].text.strip())  # Tests/1M pop
    row_data.append(cells[14].text.strip())  # Population
    row_data.append(cells[16].text.strip())  # Continent
    row_data.append(cells[17].text.strip())  # 1 Case every X ppl
    row_data.append(cells[18].text.strip())  # 1 Death every X ppl
    row_data.append(cells[19].text.strip())  # 1 Test every X ppl
    row_data.append(cells[20].text.strip())  # New Cases/1M pop
    row_data.append(cells[21].text.strip())  # New Deaths/1M pop


    #file.write(",".join(row_data))
    #file.write("\n")
    #the problem is here excel is finding  the numbers too big open with sheets cant etend on this comp
    file.write('"' + '","'.join(row_data) + '"')
    file.write("\n")
file.close()
```
